## Lendo tabelas com o DataFrame

```python
import pandas as pd
import openpyxl
from pathlib import Path
```

```python
pasta_atual = Path.cwd()
tabela_clientes = pasta_atual / "arquivos .py" / 'clientes.xlsx'

df = pd.read_excel(tabela_clientes)

print(df.head(10))
```

## Lendo planilha específica

```python
df = pd.read_excel(tabela_clientes, sheet_name='SC')

print(df.head(10))
```

# Modificando o Header

Por padrão é 0 (zero)

```python
df = pd.read_excel(tabela_clientes, sheet_name='SC', header=1)

print(df.head(10))
```

## Adicionando a coluna de index

```python
df = pd.read_excel(tabela_clientes, sheet_name='SC', index_col=0)

print(df.head(5))
```

## Lendo colunas específicas

```python
df = pd.read_excel(tabela_clientes, sheet_name='SC', usecols="A:B")

print(df.head(5))
```

## Usando ponto como separador decimal

O parâmetro 'thousands' indica o separador de milhar.

```python
df = pd.read_excel(tabela_clientes, sheet_name='SC', decimal='.', thousands=',')

print(df.head(5))
```

## Escrevendo Planilha

```python
pasta_atual = Path.cwd()

df = pd.read_excel(tabela_clientes, sheet_name='SC')

copia = df.to_excel(pasta_atual / 'arquivos .py' / 'copia_clientes.xlsx')
```

## Escrevendo diversas abas

```python
df_sc = pd.read_excel(tabela_clientes, sheet_name='SC')
df_rs = pd.read_excel(tabela_clientes, sheet_name='RS')

with pd.ExcelWriter(pasta_atual / 'arquivos .py' / 'copia_clientes.xlsx') as writer:
    df_sc.to_excel(writer, sheet_name='SC', index=False)
    df_rs.to_excel(writer, sheet_name='RS', index=False)
```
